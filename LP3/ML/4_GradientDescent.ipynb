```python
from pathlib import Path

import matplotlib.pyplot as plt
```

```python
def f(x):
    return (x + 5) ** 2


def gradient(x):
    return 2 * (x + 5)
```

```python
learning_rate = 0.1
path = [10.0]
```

```python
for _ in range(1000):
    x = path[-1]
    if abs(gradient(x)) < 1e-6:
        break
    path.append(x - learning_rate * gradient(x))
```

```python
print(f"Minimum at x = {x:.4f}, f(x) = {f(x):.4f}, after {len(path) - 1} steps")
```

```python
curve = [i / 10 for i in range(-200, 101)]
plt.figure()
plt.plot(curve, [f(c) for c in curve])
plt.plot(path, [f(p) for p in path], "ro-")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.savefig(Path.cwd() / "gradient_descent.png")
plt.show()
```
